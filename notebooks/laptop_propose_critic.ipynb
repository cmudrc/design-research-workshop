```python
import json

from design_research_workshop import laptop_propose_critic, workshop_client

battery_hours = 7  # The catalog offers 6 or 8 hours near this limit.
max_weight_lb = 3
task = (
    "Choose a student laptop configuration as JSON with "
    f"at least {battery_hours} hours of battery life and "
    f"a maximum weight of {max_weight_lb} pounds."
)

with workshop_client() as client:
    pattern = laptop_propose_critic(client)
    result = pattern.run(task, request_id="workshop-laptop-revision")

assert result.success and result.approved, result.terminated_reason
```

```python
for iteration in result.critique_iterations:
    proposal = json.loads(iteration["proposal"])
    print(f"Iteration {iteration['iteration']}")
    print(f"  Battery life: {proposal['z3']:g} hours")
    print(f"  Weight: {proposal['z4']:g} pounds")
    print(f"  Critic: {iteration['feedback'] or 'Approved'}")
```
